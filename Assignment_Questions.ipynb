## Question 1: Multi-level Linguistic Analysis

Consider the sentence: "The company's CEO didn't respond to our meeting invitation."

Analyze this sentence from four different linguistic perspectives:

- **Semantics**: Explain the meaning and relationships between words

````python
import spacy 
import nltk
from spacy import displacy
````

````python
nlp = spacy.load("en_core_web_sm")
````

````python
# The sentence to analyze
sentence = "The company's CEO didn't respond to our meeting invitation."
doc = nlp(sentence)

print(f"Analyzing: {sentence}")
````

````python
#Part-of-Speech Tagging
print("=" * 70)
print("PART-OF-SPEECH TAGS")
print("=" * 70)
print(f"{'Token':<20} {'POS':<10} {'Tag':<10} {'Description'}")
print("-" * 70)
for token in doc:
    print(f"{token.text:<20} {token.pos_:<10} {token.tag_:<10} {spacy.explain(token.tag_)}")
````

````python
#Dependency Parsing
print("=" * 70)
print("DEPENDENCY PARSE TREE")
print("=" * 70)
print(f"{'Token':<20} {'Dependency':<15} {'Head':<15} {'Children'}")
print("-" * 70)
for token in doc:
    children = [child.text for child in token.children]
    print(f"{token.text:<20} {token.dep_:<15} {token.head.text:<15} {', '.join(children)}")
````

---

````python
from spacy import displacy

# Generate HTML visualization
html = displacy.render(doc, style='dep', jupyter=False)

with open("dependency_tree.html", "w", encoding="utf-8") as f:
    f.write(html)

print("Dependency tree visualization saved to 'dependency_tree.html'")
print("Open this file in your browser to view the visualization")

# Also print a text-based tree
print("\n" + "=" * 70)
print("TEXT-BASED DEPENDENCY TREE")
print("=" * 70)

def print_tree(token, indent=0):
    arrow = "└─" if indent > 0 else ""
    print("  " * indent + arrow + f" {token.text} [{token.dep_}]")
    for child in token.children:
        print_tree(child, indent + 1)

# Find the root
root = [token for token in doc if token.head == token][0]
print_tree(root)
````

````python
#Phrase Structure
print("=" * 70)
print("PHRASE STRUCTURE")
print("=" * 70)

print("\nNoun Phrases:")
for chunk in doc.noun_chunks:
    print(f"  - '{chunk.text}' (root: {chunk.root.text}, dependency: {chunk.root.dep_})")

print("\nVerbs and Auxiliaries:")
for token in doc:
    if token.pos_ in ['VERB', 'AUX']:
        print(f"  - {token.text} ({token.pos_}, tense: {token.morph.get('Tense')})")
````

````python
print("=" * 70)
print("SEMANTIC ANALYSIS - Named Entities")
print("=" * 70)
if doc.ents:
    for ent in doc.ents:
        print(f"  - {ent.text}: {ent.label_} ({spacy.explain(ent.label_)})")
else:
    print("  No named entities detected")
````

````python
#Lexical Relationships
print("=" * 70)
print("KEY CONTENT WORDS (with lemmas)")
print("=" * 70)
content_words = [token for token in doc if token.pos_ in ['NOUN', 'VERB', 'ADJ', 'ADV', 'PROPN']]
for token in content_words:
    print(f"  - {token.text:<15} ({token.pos_:<6}) → lemma: '{token.lemma_}'")
````

````python
#Morphological Features
print("=" * 70)
print("MORPHOLOGICAL FEATURES")
print("=" * 70)
print(f"{'Token':<20} {'Lemma':<15} {'Morphological Features'}")
print("-" * 70)
for token in doc:
    morph_features = str(token.morph) if token.morph else "None"
    print(f"{token.text:<20} {token.lemma_:<15} {morph_features}")
````

**Hint**: Consider whether stemming or lemmatization is more appropriate for sentiment analysis. Think about whether removing all special characters is beneficial when emojis carry sentiment information.

Text Pre-processing Pipeline for Sentiment Analysis (E-commerce Reviews)
1. Tokenization

Include it: 

Why: Breaks sentences into words, emojis, and symbols — helps the model understand each meaningful piece.

Considerations: Use a social-media-aware tokenizer (so it keeps emojis and slang words like “gonna”, “wanna”).

Order: First step — everything else (like normalization) happens on tokens.

2. Normalization

Include it: 

Why: Cleans and standardizes text so that variations mean the same thing.

What to do:

Convert emojis 😊 → words (“smiling_face”) because emojis show strong sentiment.

Replace prices and product codes (₹599, ABC123) → <PRICE> and <CODE> (not sentiment-related).

Lowercase text (to reduce duplicates).

Expand slang (“gonna” → “going to”, “u” → “you”).

Fix simple typos (“amazng” → “amazing”).

Order: After tokenization, before further filtering — so all tokens are clean and consistent.

3. Stop-word Removal

Include it selectively: 

Why: Common words like “the”, “is”, “at” usually don’t affect sentiment.

BUT: Keep important words like “not”, “never” (they change meaning: “not good” ≠ “good”).

Order: After normalization — once words are clear and standardized.

4. Stemming / Lemmatization

Prefer Lemmatization: 

Why: Lemmatization finds real dictionary forms (“better” → “good”, “was” → “be”) and keeps meaning clear.
Stemming (“loving” → “love”, but “organized” → “organ”) can distort meaning — bad for sentiment.

Considerations: Lemmatization is slower but keeps emotional words intact — important for polarity detection.

Order: After stop-word removal — only process meaningful tokens.


Final Order of Operations
Tokenize → split text properly
Normalize → clean slang, emojis, prices, typos
Remove Stop-words (selectively) → keep negations
Lemmatize → get base, meaningful words

Why the order matters
You must tokenize first to manipulate words easily.
Normalize early to make tokens consistent.
Stop-word removal later ensures you don’t delete something important by mistake.
Lemmatize last to make final clean, meaningful words ready for modeling.

Apply both stemming (Porter Stemmer) and lemmatization to these sentences. Then:

- Discuss scenarios where stemming would be preferred over lemmatization and vice versa
- Analyze the impact on: search engines, text classification, and information retrieval systems

**Hint**: Consider computational cost, accuracy, and preservation of meaning. Words like "better", "organizing", and "was" behave differently under stemming vs lemmatization.

Sentences:

“The meeting was well organized and the organizers did a great job.”
“She is better at organizing than her predecessor was.”

1. Porter Stemming Output

“The meet wa well organ and the organ did a great job.”
“She is better at organ than her predecess wa.”

“The meeting be well organize and the organizer do a great job.”

````python
import nltk
nltk_data = ['punkt', 'averaged_perceptron_tagger']
for d in nltk_data:
    try:
        nltk.data.find('tokenizers/' + d)
    except LookupError:
        nltk.download(d)

print('Setup complete.')
````

````python
# Define the ambiguous sentences
sentences = [
    "The duck is ready to eat.",
    "They can fish.",
    "Time flies like an arrow."
]

for s in sentences:
    print(s)
````

````python
from nltk import word_tokenize, pos_tag

for s in sentences:
    tokens = word_tokenize(s)
    tags = pos_tag(tokens)
    print(f"\nSentence: {s}")
    print(tags)
````

## Rule-based vs Probabilistic Tagging

**Rule-based Tagging:**
- Uses manually defined grammar rules (e.g., determiner + noun).
- Strength: interpretable and deterministic.
- Weakness: struggles with unseen or ambiguous structures.

**Probabilistic Tagging (e.g., Hidden Markov Model):**
- Learns tag probabilities from corpora.
- Strength: handles context well.
- Weakness: depends on training data quality.

---

````python
# Simple rule-based tagging example using RegexpTagger
patterns = [
    (r'.*ing$', 'VBG'),  # Gerunds
    (r'.*ed$', 'VBD'),   # Past tense verbs
    (r'.*es$', 'VBZ'),   # 3rd person singular verbs
    (r'^-?[0-9]+(.[0-9]+)?$', 'CD'),  # Numbers
    (r'.*', 'NN')        # Default noun
]

from nltk import RegexpTagger
rule_tagger = RegexpTagger(patterns)

for s in sentences:
    tokens = word_tokenize(s)
    print(f"\nRule-based tags for: {s}")
    print(rule_tagger.tag(tokens))
````

### ✅ Summary
- POS tagging assigns grammatical roles that reduce ambiguity.
- Rule-based = simple but rigid.
- Probabilistic = flexible, context-aware, better for real-world language.
- Both have limits; combining statistical and rule-based approaches often works best.
